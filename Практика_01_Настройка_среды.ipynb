### Три способа работать с Python

Одну и ту же программу на Питоне можно запустить по-разному:

1. **Облако (Google Colab)** — код исполняется на серверах Google, ничего устанавливать не нужно. Так мы работаем, когда локальной установки нет под рукой. ⚠️ В Китае Colab напрямую недоступен — локальные инструменты из этой практики решают эту проблему, а из облачных аналогов без VPN работает [Baidu AI Studio](https://aistudio.baidu.com/) (免费 GPU-блокноты).
2. **Скрипт `.py`** — программа хранится в файле и передаётся интерпретатору из командной строки. Так мы напишем и запустим первый файл в PyCharm.
3. **Jupyter Notebook** — локальный аналог Colab: интерактивные ячейки кода, текст и графики в одном документе. Основной инструмент всех практик курса.

### Экосистема Anaconda

**Anaconda** — это дистрибутив Python и репозиторий пакетов, специально предназначенных для анализа данных и машинного обучения. Основу дистрибутива составляет система управления пакетами и окружениями **conda**. Управлять conda можно двумя способами: через **Anaconda Prompt** (аналог командной строки Windows) и через **Anaconda Navigator** (графический интерфейс).

В состав дистрибутива входят программы, которыми мы будем пользоваться весь семестр: **Jupyter Notebook и JupyterLab** — интерактивные блокноты; **Spyder и PyCharm** — интегрированные среды разработки (IDE). IDE — это редактор кода с автодополнением, анализом ошибок, отладчиком и запуском проектов одним нажатием.

<div style='text-align: center; margin: 20px;'>
<i style="color:#555;">Структура дистрибутива Anaconda: conda,.Navigator, Jupyter, IDE (источник: dmitrymakarov.ru)</i><br>
</div>

> 💡 **Почему Anaconda, а не «чистой» Python с python.org?** В Anaconda уже предустановлены NumPy, Pandas, > Matplotlib, Jupyter и ещё ~250 библиотек курса, а conda корректно ставит «тяжёлые» пакеты с бинарными > зависимостями. Кроме того, Anaconda безболезненно уживается с уже установленным Python — об этом в Шаге 3.

## ⚙️ Шаг 3. Anaconda Prompt: conda-окружение ai_course и зеркала (15 мин)

### 3.1. Что такое conda и зачем окружения

Программа **conda** объединяет менеджер пакетов (как `pip`) и систему **виртуальных окружений**. Идея окружения: если в проекте нужна строго определённая версия библиотеки, а другая её версия сломает код, правильное решение — **изолировать** набор библиотек проекта в отдельной папке. Для каждой учебной задачи можно завести своё окружение со своими версиями — они не будут конфликтовать.

По умолчанию вы работаете в **базовом окружении** (`base`). Все команды этого шага выполняются в **Anaconda Prompt** (меню «Пуск» → Anaconda Prompt) — в обычном `cmd` conda недоступна, потому что мы намеренно не ставили галочку PATH при установке.

### 3.2. Посмотрите пакеты и окружения

```bash

conda info --envs     # список всех окружений
```

### 3.4. Настройте зеркала Tsinghua для conda и pip (без VPN)

По умолчанию conda и pip скачивают пакеты с зарубежных серверов (`repo.anaconda.com`, `pypi.org`), которые из Китая работают медленно или обрываются. Одна команда переключает pip на зеркало Tsinghua:

```bash
conda activate ai_course

```

Для conda создайте файл настроек `.condarc` в домашней папке (`C:\Users\<имя>\.condarc`) со строками ниже — или выполните команды по очереди:

```bash
conda config --add channels https://mirrors.tuna.tsinghua.edu.cn/anaconda/pkgs/main/
conda config --add channels https://mirrors.tuna.tsinghua.edu.cn/anaconda/pkgs/free/
conda config --set show_channel_urls yes
```

> 💡 **Проверка:** введите `conda config --show channels` и `pip config list`. Если увидите адреса > `mirrors.tuna.tsinghua.edu.cn` — зеркала работают, и установка пойдёт со скоростью локальной сети.

## 📓 Шаг 4. Работа с Jupyter Notebook: ячейки, ядро, Markdown (10 мин)

### 4.1. Вкладка Cell: управление запуском ячеек

В верхнем меню блокнота важнейшие вкладки — **Cell** и **Kernel**.

<div style='text-align: center; margin: 20px;'>
<i style="color:#555;">Вкладка Cell: запуск и очистка вывода (источник: dmitrymakarov.ru)</i><br>
</div>

Полезные команды:

- **Run Cells** — исполнить текущую ячейку;
- **Run All** — исполнить все ячейки блокнота сверху вниз (так проверяют, что блокнот «чистый»);
- **Run All Above / Run All Below** — исполнить всё выше или ниже текущей ячейки;
- **All Output → Clear** — очистить вывод всех ячеек (полезно перед сдачей работы).

### 4.2. Вкладка Kernel: вычислительный «движок»

**Ядро (kernel)** — это процесс, который исполняет код ваших ячеек. Команды вкладки Kernel:

<div style='text-align: center; margin: 20px;'>
<i style="color:#555;">Вкладка Kernel: Interrupt и Restart (источник: dmitrymakarov.ru)</i><br>
</div>

- **Interrupt** — прервать зациклившуюся или зависшую ячейку;
- **Restart** — перезапустить ядро: все переменные стираются, но код остаётся;
- **Restart & Clear Output** — перезапуск + очистка вывода;
- **Restart & Run All** — перезапуск + полное повторное исполнение (самая честная проверка).

### 4.3. Как устроен Jupyter изнутри

<div style='text-align: center; margin: 20px;'>
<i style="color:#555;">Схема: браузер → сервер → ядро (источник: dmitrymakarov.ru)</i><br>

Пользователь работает в **браузере**; браузер отправляет запросы **серверу** — он загружает блокнот и сохраняет изменения в файле формата JSON с расширением `.ipynb`; когда нужно исполнить код, сервер обращается к **ядру**. ### 6.1. Переменные и типы данных

Переменная — имя для значения. Python сам определяет тип: строка (`str`), целое число (`int`), дробное (`float`), логическое (`bool`). f-строка подставляет значения переменных прямо в текст.

````python
# Переменные и типы данных — выполните ячейку
name = 'Студент'          # str   — строка
age = 20                  # int   — целое число
gpa = 4.5                 # float — дробное число
is_ready = True           # bool  — истина/ложь

print(f'{name}, {age} лет, средний балл {gpa}, готов к занятию: {is_ready}')
print(type(name), type(age), type(gpa), type(is_ready))
````

### 6.2. Список и словарь

**Список** хранит упорядоченный набор значений, **словарь** — пары «ключ → значение». Именно словарями и списками описываются данные для машинного обучения (столбцы таблицы, признаки объекта).

````python
# Список и словарь — выполните ячейку
grades = [85, 92, 78, 95]                      # список оценок
print('Оценки:', grades)
print('Средняя:', sum(grades) / len(grades))   # 87.5

student = {'имя': 'Ли Вэй', 'группа': 'ИИ-101', 'оценки': grades}
for key, value in student.items():
    print(f'  {key}: {value}')
````

### 6.3. Ветвление, цикл и функция

Функция — переиспользуемый блок кода с именем и параметрами. Классический учебный пример — **линейный поиск**: перебираем элементы списка, пока не найдём нужное число. Если цель найдена, возвращаем её индекс, иначе `-1`.

````python
# Функция линейного поиска — выполните ячейку
def linear_search(arr, target):
    """Возвращает индекс target в списке arr или -1, если число не найдено."""
    for i in range(len(arr)):        # цикл по позициям списка
        if arr[i] == target:         # ветвление: сравнение
            return i                 # нашёл — вернуть позицию
    return -1                        # не нашёл

numbers = [4, 7, 2, 9, 1, 8]
for t in [9, 5]:
    idx = linear_search(numbers, t)
    if idx != -1:
        print(f'Число {t} найдено на позиции {idx}')
    else:
        print(f'Число {t} в списке отсутствует')
````
